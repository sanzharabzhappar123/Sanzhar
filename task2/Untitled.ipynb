```python
# Importing Pandas library
import pandas as pd

# Create a DataFrame with columns: Name, Age, Score
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [23, 25, 22, 24, 23],
    'Score': [85, 90, 88, 92, 78]
}
df = pd.DataFrame(data)
print("DataFrame:")
print(df)
```

```python
# Display the first 3 rows
print("First 3 rows:")
print(df.head(3))

# Show summary statistics using describe()
print("Summary statistics:")
print(df.describe())

# Find the average score
average_score = df['Score'].mean()
print("Average score:", average_score)
```

```python
# Select students older than 20
older_than_20 = df[df['Age'] > 20]
print("Students older than 20:")
print(older_than_20)

# Select students with a score above 80
above_80 = df[df['Score'] > 80]
print("Students with a score above 80:")
print(above_80)
```
